```python
#exp2 :finite state machine
def accepts(s):
    state = 0

    for ch in s:
        if state == 0:
            if ch == 'a':
                state = 1
            else:
                state = 0

        elif state == 1:
            if ch == 'b':
                state = 2
            elif ch == 'a':
                state = 1
            else:
                state = 0

        elif state == 2:
            if ch == 'a':
                state = 1
            else:
                state = 0

    return state == 2


s = input("Enter a string: ")

if accepts(s):
    print("Accepted")
else:
    print("Rejected")
```

```python
# 4.Finite State Machine for Morphological Parsing
# Generates plural forms of English nouns


def generate_plural(noun):

    # State q0 (Start State)
    if noun.endswith(("s", "x", "z", "ch", "sh")):
        # Transition to q1
        plural = noun + "es"

    elif noun.endswith("y") and len(noun) > 1 and noun[-2].lower() not in "aeiou":
        # Transition to q2
        plural = noun[:-1] + "ies"

    else:
        # Transition to q3
        plural = noun + "s"

    return plural


# Test words
words = ["cat", "dog", "bus", "box", "church",
         "dish", "baby", "city", "toy", "book"]

print("Singular\tPlural")
print("------------------------")

for word in words:
   print(f"{word:10}\t{generate_plural(word)}")
```

```python
#exp 6 :
import random

text = "I like Python. I like machine learning. I like coding."

words = text.split()

bigrams = {}

for i in range(len(words) - 1):
    word = words[i]
    next_word = words[i + 1]

    if word not in bigrams:
        bigrams[word] = []

    bigrams[word].append(next_word)

word = "I"
result = [word]

for i in range(10):
    if word in bigrams:
        word = random.choice(bigrams[word])
        result.append(word)
    else:
        break

print("Generated text:")
print(" ".join(result))
```

```python
#exp 8 :
words = ["I", "like", "dogs"]

probability = {
    "I": {"PRP": 0.9, "NN": 0.1},
    "like": {"VB": 0.7, "IN": 0.3},
    "dogs": {"NNS": 0.8, "VB": 0.2}
}

for word in words:
    tag = max(probability[word], key=probability[word].get)
    print(word, "->", tag)
```

```python
#EXP-11
# Top-Down Parser

grammar = {
    "S": [["NP", "VP"]],
    "NP": [["Det", "N"]],
    "VP": [["V"]],
    "Det": [["the"]],
    "N": [["boy"], ["girl"]],
    "V": [["runs"], ["plays"]]
}

def parse(symbol, words, pos):
    if symbol not in grammar:
        if pos < len(words) and symbol == words[pos]:
            return pos + 1
        return None

    for rule in grammar[symbol]:
        current = pos

        for part in rule:
            result = parse(part, words, current)

            if result is None:
                break

            current = result
        else:
            return current

    return None


sentence = input("Enter sentence: ").lower()
words = sentence.split()

result = parse("S", words, 0)

if result == len(words):
    print("Sentence Accepted")
else:
    print("Sentence Rejected")
```

```python
#EXP 12
# Simple Earley Parser

grammar = {
    "S": [["NP", "VP"]],
    "NP": [["Det", "N"]],
    "VP": [["V"]],
    "Det": [["the"]],
    "N": [["boy"], ["girl"]],
    "V": [["runs"], ["plays"]]
}


def earley_parse(words):

    n = len(words)

    # Chart contains states for each position
    chart = [[] for _ in range(n + 1)]

    # State format:
    # (left side, right side, position, start position)
    chart[0].append(("S'", ["S"], 0, 0))

    for i in range(n + 1):

        changed = True

        while changed:
            changed = False

            for state in chart[i]:
                left, right, dot, start = state

                # Prediction
                if dot < len(right):
                    symbol = right[dot]

                    if symbol in grammar:
                        for rule in grammar[symbol]:
                            new_state = (symbol, rule, 0, i)

                            if new_state not in chart[i]:
                                chart[i].append(new_state)
                                changed = True

                # Completion
                else:
                    for old in chart[start]:
                        l, r, d, s = old

                        if d < len(r) and r[d] == left:
                            new_state = (l, r, d + 1, s)

                            if new_state not in chart[i]:
                                chart[i].append(new_state)
                                changed = True

        # Scanning
        if i < n:
            for state in chart[i]:

                left, right, dot, start = state

                if dot < len(right):
                    symbol = right[dot]

                    if symbol == words[i]:

                        new_state = (
                            left,
                            right,
                            dot + 1,
                            start
                        )

                        if new_state not in chart[i + 1]:
                            chart[i + 1].append(new_state)

    # Check final state
    final_state = ("S'", ["S"], 1, 0)

    return final_state in chart[n]


sentence = input("Enter sentence: ").lower()
words = sentence.split()

if earley_parse(words):
    print("Sentence Accepted")
else:
    print("Sentence Rejected")
```

```python
#exp-13 :
from nltk import CFG
from nltk.parse import EarleyChartParser
grammer=CFG.fromstring("""
 S -> NP VP
 NP -> DET N
 VP -> V NP
 DET -> 'the'
 N -> 'cat'|'dog'
 V -> 'chased'
""")
parser = EarleyChartParser(grammer)

sentence = 'THE CAT CHASED THE DOG'.lower().split()
trees = __builtins__.list(parser.parse(sentence))

if trees:
    print("sentence is accepted")
    for tree in trees:
        print(tree)
else:
    print("sentence is rejected")
```

```python
#EXP-15
import nltk
from nltk  import PCFG
from nltk.parse import ViterbiParser

grammer=PCFG.fromstring("""
S -> NP VP [1.0]
NP -> Det N [0.6]
NP -> 'students'[0.4]
VP -> V NP [0.5]
VP -> V [0.5]
Det -> 'the' [1.0]
N -> 'students'[1.0]
V -> 'study' [1.0]

""")

parser = ViterbiParser(grammer)

sentence = " the students study ".lower().split()
trees = __builtins__.list(parser.parse(sentence))

for tree in trees :
  print(tree)
  print(tree.pretty_print())
  print(tree.prob())

```
