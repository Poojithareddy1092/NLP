```python
#exp2 :finite state machine
def accepts(s):
    state = 0

    for ch in s:
        if state == 0:
            if ch == 'a':
                state = 1
            else:
                state = 0

        elif state == 1:
            if ch == 'b':
                state = 2
            elif ch == 'a':
                state = 1
            else:
                state = 0

        elif state == 2:
            if ch == 'a':
                state = 1
            else:
                state = 0

    return state == 2


s = input("Enter a string: ")

if accepts(s):
    print("Accepted")
else:
    print("Rejected")
```

```python
# 4.Finite State Machine for Morphological Parsing
# Generates plural forms of English nouns


def generate_plural(noun):

    # State q0 (Start State)
    if noun.endswith(("s", "x", "z", "ch", "sh")):
        # Transition to q1
        plural = noun + "es"

    elif noun.endswith("y") and len(noun) > 1 and noun[-2].lower() not in "aeiou":
        # Transition to q2
        plural = noun[:-1] + "ies"

    else:
        # Transition to q3
        plural = noun + "s"

    return plural


# Test words
words = ["cat", "dog", "bus", "box", "church",
         "dish", "baby", "city", "toy", "book"]

print("Singular\tPlural")
print("------------------------")

for word in words:
   print(f"{word:10}\t{generate_plural(word)}")
```

```python
#exp 6 :
import random

text = "I like Python. I like machine learning. I like coding."

words = text.split()

bigrams = {}

for i in range(len(words) - 1):
    word = words[i]
    next_word = words[i + 1]

    if word not in bigrams:
        bigrams[word] = []

    bigrams[word].append(next_word)

word = "I"
result = [word]

for i in range(10):
    if word in bigrams:
        word = random.choice(bigrams[word])
        result.append(word)
    else:
        break

print("Generated text:")
print(" ".join(result))
```

```python
#EXP-11
# Top-Down Parser

grammar = {
    "S": [["NP", "VP"]],
    "NP": [["Det", "N"]],
    "VP": [["V"]],
    "Det": [["the"]],
    "N": [["boy"], ["girl"]],
    "V": [["runs"], ["plays"]]
}

def parse(symbol, words, pos):
    if symbol not in grammar:
        if pos < len(words) and symbol == words[pos]:
            return pos + 1
        return None

    for rule in grammar[symbol]:
        current = pos

        for part in rule:
            result = parse(part, words, current)

            if result is None:
                break

            current = result
        else:
            return current

    return None


sentence = input("Enter sentence: ").lower()
words = sentence.split()

result = parse("S", words, 0)

if result == len(words):
    print("Sentence Accepted")
else:
    print("Sentence Rejected")
```

```python
#EXP 12
# Simple Earley Parser

grammar = {
    "S": [["NP", "VP"]],
    "NP": [["Det", "N"]],
    "VP": [["V"]],
    "Det": [["the"]],
    "N": [["boy"], ["girl"]],
    "V": [["runs"], ["plays"]]
}


def earley_parse(words):

    n = len(words)

    # Chart contains states for each position
    chart = [[] for _ in range(n + 1)]

    # State format:
    # (left side, right side, position, start position)
    chart[0].append(("S'", ["S"], 0, 0))

    for i in range(n + 1):

        changed = True

        while changed:
            changed = False

            for state in chart[i]:
                left, right, dot, start = state

                # Prediction
                if dot < len(right):
                    symbol = right[dot]

                    if symbol in grammar:
                        for rule in grammar[symbol]:
                            new_state = (symbol, rule, 0, i)

                            if new_state not in chart[i]:
                                chart[i].append(new_state)
                                changed = True

                # Completion
                else:
                    for old in chart[start]:
                        l, r, d, s = old

                        if d < len(r) and r[d] == left:
                            new_state = (l, r, d + 1, s)

                            if new_state not in chart[i]:
                                chart[i].append(new_state)
                                changed = True

        # Scanning
        if i < n:
            for state in chart[i]:

                left, right, dot, start = state

                if dot < len(right):
                    symbol = right[dot]

                    if symbol == words[i]:

                        new_state = (
                            left,
                            right,
                            dot + 1,
                            start
                        )

                        if new_state not in chart[i + 1]:
                            chart[i + 1].append(new_state)

    # Check final state
    final_state = ("S'", ["S"], 1, 0)

    return final_state in chart[n]


sentence = input("Enter sentence: ").lower()
words = sentence.split()

if earley_parse(words):
    print("Sentence Accepted")
else:
    print("Sentence Rejected")
```

```python
#exp-13 :
from nltk import CFG
from nltk.parse import EarleyChartParser
grammer=CFG.fromstring("""
 S -> NP VP
 NP -> DET N
 VP -> V NP
 DET -> 'the'
 N -> 'cat'|'dog'
 V -> 'chased'
""")
parser = EarleyChartParser(grammer)

sentence = 'THE CAT CHASED THE DOG'.lower().split()
trees = __builtins__.list(parser.parse(sentence))

if trees:
    print("sentence is accepted")
    for tree in trees:
        print(tree)
else:
    print("sentence is rejected")
```

```python
# 20. INFORMATION RETRIEVAL USING TF-IDF

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

documents = [
    "Artificial intelligence is transforming technology",
    "Machine learning is a branch of artificial intelligence",
    "Natural language processing deals with human language",
    "Deep learning is used in artificial intelligence applications"
]

query = input("Enter your search query: ")

vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(documents)
query_vector = vectorizer.transform([query])

similarity = cosine_similarity(query_vector, tfidf_matrix)[0]

ranking = similarity.argsort()[::-1]

print("\nDocument Ranking:")

for i in ranking:
    print(f"Document {i + 1}: Score = {similarity[i]:.4f}")
    print(documents[i])
```

```python
# 21. SYNTAX-DRIVEN SEMANTIC ANALYSIS

import nltk
from nltk import word_tokenize, pos_tag
from nltk.corpus import wordnet

nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("averaged_perceptron_tagger")
nltk.download("averaged_perceptron_tagger_eng")
nltk.download("wordnet")
nltk.download("omw-1.4")

sentence = input("Enter a sentence: ")

words = word_tokenize(sentence)
tags = pos_tag(words)

noun_phrases = []

for i in range(len(tags) - 1):
    if tags[i][1] in ["DT", "JJ"] and tags[i + 1][1].startswith("NN"):
        noun_phrases.append(
            tags[i][0] + " " + tags[i + 1][0]
        )

for word, tag in tags:
    if tag.startswith("NN"):
        noun_phrases.append(word)

noun_phrases = list(dict.fromkeys(noun_phrases))

print("\nNoun Phrases and Meanings:")

for phrase in noun_phrases:

    noun = phrase.split()[-1]

    meanings = wordnet.synsets(noun)

    print("\nNoun Phrase:", phrase)

    if meanings:
        print("Meaning:", meanings[0].definition())
    else:
        print("Meaning not found")
```

```python
# 22. REFERENCE RESOLUTION

import re

text = input("Enter a text: ")

sentences = re.split(r'[.!?]', text)

male_names = {"ravi", "arun", "rahul", "raj"}
female_names = {"sita", "anita", "priya"}

last_male = None
last_female = None

for sentence in sentences:

    words = sentence.lower().split()

    for word in words:

        if word in male_names:
            last_male = word

        elif word in female_names:
            last_female = word

        elif word in ["he", "him", "his"]:
            print(word, "->", last_male)

        elif word in ["she", "her", "hers"]:
            print(word, "->", last_female)
```

```python
# 23. TEXT COHERENCE EVALUATION

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

text = input("Enter a paragraph: ")

sentences = [
    s.strip()
    for s in text.split(".")
    if s.strip()
]

if len(sentences) < 2:
    print("Enter at least two sentences.")

else:

    vectorizer = TfidfVectorizer()

    vectors = vectorizer.fit_transform(sentences)

    similarities = []

    for i in range(len(sentences) - 1):

        score = cosine_similarity(
            vectors[i],
            vectors[i + 1]
        )[0][0]

        similarities.append(score)

        print(
            f"Similarity between Sentence {i + 1} "
            f"and Sentence {i + 2}: {score:.4f}"
        )

    coherence_score = sum(similarities) / len(similarities)

    print("\nOverall Coherence Score:",
          round(coherence_score, 4))

    if coherence_score >= 0.5:
        print("Text is relatively coherent.")
    else:
        print("Text has low coherence.")
```
