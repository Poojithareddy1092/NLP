```python
# C04_AT2.1

import nltk
from nltk import CFG
from nltk.parse import EarleyChartParser

grammar = CFG.fromstring("""
S -> NP VP
NP -> PRP | Det N | Det N PP
VP -> V NP | V NP PP | V NP PP PP
PP -> P NP
Det -> "the"
N -> "damage" | "car" | "accident"
V -> "reported"
PRP -> "I"
P -> "to" | "after"
""")

parser = EarleyChartParser(grammar)

sentence = "I reported the damage to the car after the accident".split()

print("Question 1 - Insurance Claim Processing")
print("\nSentence:")
print(" ".join(sentence))

print("\nPossible Parse Trees:")

for tree in parser.parse(sentence):
    print(tree)
    print()
```

```python
# C04_AT2.2

import nltk
from nltk import CFG
from nltk.parse import EarleyChartParser

grammar = CFG.fromstring("""
S -> VP
VP -> V NP PP PP
NP -> Num N
PP -> P NP
NP -> Det N
V -> "Book"
Num -> "two"
N -> "tickets" | "parents" | "sleeper"
P -> "from" | "to" | "for" | "with"
Det -> "my"
""")

parser = EarleyChartParser(grammar)

sentence = "Book two tickets from Chennai to Delhi for my parents with AC sleeper".split()

print("Question 2 - Railway Ticket Booking")

print("\nSentence:")
print(" ".join(sentence))

print("\nEarley Parse:")

try:
    for tree in parser.parse(sentence):
        print(tree)
except:
    print("Partial or ambiguous input detected.")

print("\nParser Types")
print("Top-Down Parsing: Starts from S and expands rules.")
print("CFG Parsing: Uses grammar rules to create parse trees.")
print("Earley Parsing: Handles ambiguity and partial input.")
```

```python
# C04_AT2.3

import nltk
from nltk import CFG
from nltk.parse import EarleyChartParser

grammar = CFG.fromstring("""
S -> NP VP
NP -> Det N
NP -> Det N Rel
Rel -> Pro V NP PP
VP -> V NP PP
PP -> P NP
Det -> "the"
N -> "supplier" | "equipment" | "company" | "components"
Pro -> "who"
V -> "delivered" | "must" | "replace"
P -> "to" | "within"
""")

parser = EarleyChartParser(grammar)

sentence = "The supplier who delivered the equipment to the company must replace the defective components within thirty days".split()

print("Question 3 - Legal Contract Analysis")

print("\nSentence:")
print(" ".join(sentence))

print("\nEarley Parse:")

try:
    for tree in parser.parse(sentence):
        print(tree)
except:
    print("Complex sentence detected.")

print("\nExtracted Information")
print("Supplier    : supplier")
print("Action      : replace")
print("Equipment   : equipment")
print("Company     : company")
print("Obligation  : must replace")
print("Deadline    : within thirty days")
```

```python
# C04_AT2.4

import nltk
from nltk import CFG
from nltk.parse import EarleyChartParser

grammar = CFG.fromstring("""
S -> V NP PP PP PP
NP -> N
PP -> P NP
NP -> Det N
N -> "laptops" | "touchscreen" | "students"
V -> "Find"
P -> "with" | "for" | "under"
Det -> "a"
""")

parser = EarleyChartParser(grammar)

sentence = "Find laptops with a touchscreen for students under 60000".split()

print("Question 4 - E-Commerce Product Search")

print("\nQuery:")
print(" ".join(sentence))

print("\nPossible Parse:")

try:
    for tree in parser.parse(sentence):
        print(tree)
except:
    print("Ambiguous or incomplete query.")

print("\nExtracted Information")
print("Product    : laptops")
print("Feature    : touchscreen")
print("Users      : students")
print("Price      : under 60000")

print("\nParsing Methods")
print("CFG: Basic grammar-based parsing")
print("PCFG: Uses probabilities to resolve ambiguity")
print("Feature Structures: Store product attributes")
print("Earley: Handles ambiguous and incomplete input")
```

```python
# C04_AT2.5

import nltk
from nltk import CFG
from nltk.parse import EarleyChartParser

grammar = CFG.fromstring("""
S -> VP
VP -> V NP PP PP
NP -> Det N | Num N
PP -> P NP
V -> "Change"
Det -> "my"
Num -> "two"
N -> "flight" | "passengers" | "Mumbai" | "Bangalore"
P -> "to" | "with"
""")

parser = EarleyChartParser(grammar)

sentence1 = "Change my flight to Mumbai tomorrow with two passengers".split()

print("Question 5 - Airline Voice Assistant")

print("\nCommand 1:")
print(" ".join(sentence1))

print("\nEarley Parse:")

try:
    for tree in parser.parse(sentence1):
        print(tree)
except:
    print("Incomplete or ambiguous command.")

print("\nStructured Request")
print("Action       : Change")
print("Flight       : my flight")
print("Destination  : Mumbai")
print("Date         : tomorrow")
print("Passengers   : 2")

sentence2 = "Change my flight to Mumbai tomorrow actually make that Bangalore".split()

print("\nCommand 2:")
print(" ".join(sentence2))

print("\nCorrection Handling")
print("Initial Destination : Mumbai")
print("Corrected Destination: Bangalore")

print("\nProcessing Workflow")
print("Speech")
print("  -> Text")
print("  -> Tokenization")
print("  -> Earley Parsing")
print("  -> PCFG Ambiguity Resolution")
print("  -> Feature Structures")
print("  -> Booking Request")
```
